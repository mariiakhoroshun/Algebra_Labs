### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```sage
from collections import Counter

groups = {
    "Z_400": AbelianGroup([400]),
    "Z_16 x Z_25": AbelianGroup([16, 25]),
    "Z_20 x Z_20": AbelianGroup([20, 20]),
    "Z_4 x Z_100": AbelianGroup([4, 100]),
    "Z_2 x Z_10 x Z_20": AbelianGroup([2, 10, 20]),
    "Z*_505": Integers(505).unit_group()
}

print("Групи ініціалізовано:", list(groups.keys()))
```

```sage
for name, G in groups.items():
    inv = G.elementary_divisors()
    prime_powers = sorted(
        [(p, k, p^k) for m in inv for p, k in factor(m)],
        key=lambda item: (item[0], item[1])
    )
    ed = tuple(item[2] for item in prime_powers)
    p_decomp = " × ".join(f"Z_{d}" for d in ed)
    
    print(f"=== {name} (порядок = {G.order()}) ===")
    print(f"1. Елементарні дільники (p^k): {ed}")
    print(f"   Інваріантні дільники (m_i): {inv}")
    print(f"2. Добуток циклічних p-груп: {p_decomp}\n")
```

```sage
for name, G in groups.items():
    order_counts = sorted(Counter(g.order() for g in G).items())
    orders_str = ", ".join(f"ord {k}: {v}" for k, v in order_counts)
    
    print(f"=== {name} ===")
    print(f"3. Кількість елементів за порядками:\n   {orders_str}\n")
```

```sage
print("=== 4. Класи ізоморфних груп ===")
names = list(groups.keys())
visited = set()

for i in range(len(names)):
    if names[i] in visited:
        continue
    iso_class = [names[i]]
    visited.add(names[i])
    for j in range(i + 1, len(names)):
        if groups[names[i]].is_isomorphic(groups[names[j]]):
            iso_class.append(names[j])
            visited.add(names[j])
    print(" ≅ ".join(iso_class))
```

### 2. Одна абелева група порядку $n$
1. Реалізуйте функцію, яка для заданого натурального числа $n$ обчислює кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```sage
def count_abelian_groups(n):
    if n < 1:
        raise ValueError("n має бути натуральним числом")
    return prod(number_of_partitions(k) for p, k in factor(n))

for test_n in [1, 6, 16, 144, 400]:
    print(f"n = {test_n} ({factor(test_n)}): кількість груп = {count_abelian_groups(test_n)}")
```

```sage
single_group_ns = []
candidate = 1

while len(single_group_ns) < 100:
    if count_abelian_groups(candidate) == 1:
        single_group_ns.append(candidate)
    candidate += 1

print("Перші 100 чисел n, для яких існує рівно одна абелева група:")
print(single_group_ns)

print("\nРозклад на прості множники для кожного зі 100 знайдених n:")
for idx in range(0, 100, 4):
    row = [f"{n:>3} = {str(factor(n)):<12}" for n in single_group_ns[idx:idx+4]]
    print(" | ".join(row))

all_exponents = [k for n in single_group_ns for p, k in factor(n)]
print("\nАналіз показників степенів у розкладах:")
print(f"Множина всіх показників k_i, що зустрілися: {set(all_exponents)}")
print(f"Усі знайдені числа вільні від квадратів (is_squarefree): {all(is_squarefree(n) for n in single_group_ns)}")
```

```sage
N_max = 10000
mismatches = [
    n for n in range(1, N_max + 1)
    if (count_abelian_groups(n) == 1) != is_squarefree(n)
]

print(f"Перевірено чисел від 1 до {N_max}.")
print(f"Кількість розбіжностей із гіпотезою: {len(mismatches)}")

G1 = AbelianGroup([4])
G2 = AbelianGroup([2, 2])
print("\nКонтрприклад для n = 4 (не вільне від квадратів):")
print(f"Існують дві групи: Z_4 та Z_2 × Z_2. Чи ізоморфні вони? -> {G1.is_isomorphic(G2)}")
```

### 3. Дільники нуля та одиниці в скінченних кільцях
Розгляньте скінченні кільця $\mathbb{Z}_{30}$, $\mathbb{Z}_{12}\times\mathbb{Z}_{15}$, $M_3(\mathbb{Z}_3)$ та для кожного з них виконайте наступні кроки:
1. Знайдіть дільники нуля та їх кількість.
2. Знайдіть дільники одиниці та їх кількість.
3. Порівняйте отримані результати та зробіть висновок про взаємозв'язок між дільниками нуля та дільниками одиниці в скінченних кільцях.


```sage
R1 = Integers(30)
elems1 = list(R1)

zd_1 = [x for x in elems1 if x != R1.zero() and any(x * y == R1.zero() for y in elems1 if y != R1.zero())]

units_1 = [x for x in elems1 if any(x * y == R1.one() for y in elems1)]

print(f"=== Кільце Z_30 (усього елементів: {len(elems1)}) ===")
print(f"1. Дільники нуля (кількість = {len(zd_1)}):\n   {zd_1}")
print(f"2. Дільники одиниці (кількість = {len(units_1)}):\n   {units_1}")
```

```sage
R2 = cartesian_product([Integers(12), Integers(15)])
elems2 = list(R2)

zd_2 = [x for x in elems2 if x != R2.zero() and any(x * y == R2.zero() for y in elems2 if y != R2.zero())]
units_2 = [x for x in elems2 if any(x * y == R2.one() for y in elems2)]

print(f"=== Кільце Z_12 x Z_15 (усього елементів: {len(elems2)}) ===")
print(f"1. Дільники нуля (кількість = {len(zd_2)}):")
print(f"   Перші 12 прикладів: {zd_2[:12]}")
print(f"2. Дільники одиниці (кількість = {len(units_2)}):\n   {units_2}")
```

```sage
M = MatrixSpace(GF(3), 3, 3)

units_3 = []
zd_3 = []

for A in M:
    if A == M.zero():
        continue
    if A.is_invertible():
        units_3.append(A)
    else:
        zd_3.append(A)

print(f"=== Кільце M_3(Z_3) (усього матриць: {M.cardinality()}) ===")
print(f"1. Кількість дільників нуля (det A = 0, A != O): {len(zd_3)}")
print(f"2. Кількість дільників одиниці (det A != 0): {len(units_3)}")

sample_zd = zd_3[0]
v = sample_zd.right_kernel().basis()[0]
witness_B = M([[v[0], 0, 0], [v[1], 0, 0], [v[2], 0, 0]])

print("\nПриклад дільника нуля A:")
print(sample_zd)
print("Ненульова матриця B, для якої A * B = O:")
print(witness_B)
print("Перевірка A * B == O:", sample_zd * witness_B == M.zero())

print("\nПриклад дільника одиниці U та оберненої до неї U^(-1):")
sample_unit = units_3[1]
print(sample_unit)
print("Перевірка U * U^(-1) == E:", sample_unit * sample_unit.inverse() == M.one())
```

```sage
summary = [
    ("Z_30", len(elems1), len(zd_1), len(units_1)),
    ("Z_12 x Z_15", len(elems2), len(zd_2), len(units_2)),
    ("M_3(Z_3)", M.cardinality(), len(zd_3), len(units_3)),
]

print(f"{'Кільце R':<15} | {'|R|':<8} | {'Дільники нуля':<15} | {'Одиниці |R*|':<14} | {'1 + |ZD| + |R*| == |R|'}")
print("-" * 82)
for name, total, n_zd, n_u in summary:
    check = (1 + n_zd + n_u == total)
    print(f"{name:<15} | {total:<8} | {n_zd:<15} | {n_u:<14} | {check}")
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```sage
R_Q.<x> = PolynomialRing(QQ)
f_Q = x^4 + x + 2
I_Q = R_Q.ideal(f_Q)
Q_Q = R_Q.quotient(I_Q, 'xbar')
xbar_Q = Q_Q.gen()

print("=== 1. Кільце R = Q ===")
print("1. Ідеал I у Q[x]:", I_Q)
print("2. Факторкільце Q[x]/I:", Q_Q)
print("   Порядок факторкільця:", Q_Q.cardinality())
print("3. Чи є Q[x]/I полем?:", Q_Q.is_field())
print("4. Дільники нуля у Q[x]/I: відсутні (кількість = 0), оскільки I — максимальний ідеал:", I_Q.is_maximal())
print("5. Розклад f(x) над Q:", f_Q.factor(), "-> лінійних множників немає (f незвідний над Q)")
print("6. Чи оборотний елемент x + I?:", xbar_Q.is_unit())
if xbar_Q.is_unit():
    inv_Q = xbar_Q^(-1)
    print("   Обернений елемент (x + I)^(-1):", inv_Q)
    print("   Перевірка (x + I) * (x + I)^(-1) == 1:", xbar_Q * inv_Q == Q_Q.one())
```

```sage
R_4.<x> = PolynomialRing(Integers(4))
f_4 = x^4 + x + 2
I_4 = R_4.ideal(f_4)
Q_4 = R_4.quotient(I_4, 'xbar')
xbar_4 = Q_4.gen()

elems_4 = list(Q_4)

zd_4 = [a for a in elems_4 if a != Q_4.zero() and any(a * b == Q_4.zero() for b in elems_4 if b != Q_4.zero())]
units_4 = [a for a in elems_4 if any(a * b == Q_4.one() for b in elems_4)]

rem_4 = f_4
lin_factors_4 = []
for r in Integers(4):
    while rem_4(r) == 0:
        lin_factors_4.append(x - r)
        rem_4 = rem_4.quo_rem(x - r)[0]

print("=== 2. Кільце R = Z_4 ===")
print("1. Ідеал I у Z_4[x]:", I_4)
print("2. Факторкільце Z_4[x]/I:", Q_4)
print("   Порядок факторкільця:", len(elems_4))
print("3. Чи є Z_4[x]/I полем?:", len(zd_4) == 0)
print(f"4. Дільники нуля у Z_4[x]/I: кількість = {len(zd_4)} (оборотних елементів = {len(units_4)})")
print("   Приклади дільників нуля:", zd_4[:10])
print(f"5. Лінійні множники f(x) над Z_4: {lin_factors_4}, залишковий множник: ({rem_4})")
print("   Перевірка розкладу:", prod(lin_factors_4) * rem_4 == f_4)

zd_from_lin = [a for a in zd_4 if any(a.lift().quo_rem(lf)[1] == 0 for lf in lin_factors_4)]
print(f"   З {len(zd_4)} дільників нуля лише {len(zd_from_lin)} діляться на лінійні множники f(x);")
print("   решта виникають через дільник нуля 2 у самому базовому кільці Z_4 (наприклад, 2 + I).")

is_x_unit_4 = (xbar_4 in units_4)
print("6. Чи оборотний елемент x + I?:", is_x_unit_4)
if not is_x_unit_4:
    witness_4 = next(b for b in elems_4 if b != Q_4.zero() and xbar_4 * b == Q_4.zero())
    print(f"   x + I є дільником нуля! Анулятор: ({witness_4}), бо (x + I)*({witness_4}) = {xbar_4 * witness_4}")
```

```sage
R_5.<x> = PolynomialRing(GF(5))
f_5 = x^4 + x + 2
I_5 = R_5.ideal(f_5)
Q_5 = R_5.quotient(I_5, 'xbar')
xbar_5 = Q_5.gen()

elems_5 = list(Q_5)
zd_5 = []
units_5 = []
for a in elems_5:
    if a == Q_5.zero():
        continue
    g = gcd(a.lift(), f_5)
    if g == 1:
        units_5.append(a)
    else:
        b = Q_5(f_5 // g)
        assert a * b == Q_5.zero()
        zd_5.append(a)

fact_5 = f_5.factor()
lin_factors_5 = [p for p, exp in fact_5 if p.degree() == 1]

print("=== 3. Кільце R = Z_5 ===")
print("1. Ідеал I у Z_5[x]:", I_5)
print("2. Факторкільце Z_5[x]/I:", Q_5)
print("   Порядок факторкільця:", len(elems_5))
print("3. Чи є Z_5[x]/I полем?:", Q_5.is_field())
print(f"4. Дільники нуля у Z_5[x]/I: кількість = {len(zd_5)} (оборотних елементів = {len(units_5)})")
print("   Приклади дільників нуля:", zd_5[:10])
print(f"5. Повний розклад f(x) над Z_5: {fact_5}")
print(f"   Лінійний множник: {lin_factors_5[0]} (у Z_5 це те саме, що x - 2)")

zd_from_lin_5 = [a for a in zd_5 if a.lift().quo_rem(lin_factors_5[0])[1] == 0]
print(f"   Кількість дільників нуля, кратних лінійному множнику ({lin_factors_5[0]}): {len(zd_from_lin_5)}")
print(f"   Кількість дільників нуля, кратних кубічному множнику (x^3 + 2x^2 + 4x + 4): {len(zd_5) - len(zd_from_lin_5)}")

print("6. Чи оборотний елемент x + I?:", xbar_5.is_unit())
if xbar_5.is_unit():
    inv_5 = xbar_5^(-1)
    print("   Обернений елемент (x + I)^(-1):", inv_5)
    print("   Перевірка (x + I) * (x + I)^(-1) == 1:", xbar_5 * inv_5 == Q_5.one())
```

### 6*. Розфарбування кубика Рубика $2\times2\times2$

Розглянемо кубик Рубика розміру $2\times2\times2$, який має $24$ зовнішні грані кутиків (наліпки):

![rubik cube](../assets/rubik_cube.jpg)

1. Побудуйте групу поворотів кубика Рубика 2х2х2 як підгрупу симетричної групи $S_{24}$ та знайдіть її порядок.
2. Знайдіть кількість різних розфарбувань кубика у 10 кольорів. Вважайте два розфарбування однаковими, якщо одне можна отримати з іншого за допомогою поворотів кубика.



```sage
G = SymmetricGroup(24)

g1 = G('(1, 2, 4, 3)(5, 24, 9, 7)(6, 23, 10, 8)')          
g2 = G('(7, 8, 14, 13)(3, 9, 18, 12)(6, 4, 15, 17)')       
g3 = G('(9, 10, 16, 15)(4, 24, 20, 14)(8, 2, 22, 18)')     
g4 = G('(17, 18, 20, 19)(13, 15, 22, 11)(12, 14, 16, 21)') 
g5 = G('(5, 6, 12, 11)(1, 7, 17, 21)(3, 13, 19, 23)')      
g6 = G('(24, 23, 21, 22)(2, 5, 19, 16)(10, 1, 11, 20)')    

H = G.subgroup([g1, g2, g3, g4, g5, g6])
print("1. Порядок групи поворотів граней H:", H.order())
print("   Теоретична перевірка (8! * 3^7 == |H|):", factorial(8) * 3^7 == H.order())

colors = 10
fixed_sum = sum(
    cl.cardinality() * (colors ^ len(cl.representative().cycle_tuples(singletons=True)))
    for cl in H.conjugacy_classes()
)
num_colorings = fixed_sum // H.order()

print("   Кількість класів спряженості в H:", len(H.conjugacy_classes()))
print("2. Кількість різних розфарбувань у 10 кольорів:", num_colorings)
```
