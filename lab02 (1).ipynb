```sage
# ==========================================
# 3. Z_14
# ==========================================
print("=== Група Z_14 ===")
Z14 = Integers(14)                                      
print("2. Порядок:", Z14.cardinality())                 
all_gens_Z14 = [x for x in Z14 if gcd(x, 14) == 1]
print("3. Всі твірні:", all_gens_Z14)                           
print("4. Елементи:", list(Z14))                        
print("5. Таблиця Келі:\n", Z14.addition_table(names='elements'))     
print("6. Абелева?", Z14.is_commutative())              
print(f"7. Структура: C{Z14.cardinality()} (циклічна група)")
print("\n")
```

```sage
# ==========================================
# 4. Z_14^*
# ==========================================
print("=== Група Z_14^* ===")
Z14_star = Integers(14).unit_group()

print("2. Порядок:", Z14_star.order())
group_order = Z14_star.order()
all_gens_Z14_star = [int(x.value()) for x in Z14_star if x.order() == group_order]
print("3. Всі твірні:", all_gens_Z14_star)

elements = [int(x.value()) for x in Z14_star.list()]
print("4. Елементи:", elements)

print("5. Таблиця Келі (множення за модулем 14):")
print("   * | " + " ".join(f"{x:2}" for x in elements))
print("-----+-" + "-" * 3 * len(elements))
for row in elements:
    row_str = f"{row:2}   | " + " ".join(f"{(row*col)%14:2}" for col in elements)
    print(row_str)

print("6. Абелева?", Z14_star.is_abelian())
if Z14_star.is_cyclic():
    print(f"7. Структура: C{Z14_star.order()} (циклічна група)")
else:
    print(f"7. Структура: {Z14_star.structure_description()}")
print("\n")
```

### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
# ==========================================
# 1. Z_10
# ==========================================
print("=== 1. Група Z_10 ===")
Z10 = Integers(10)
a, b = Z10(2), Z10(3)

print(f"2. Довільні елементи: a = {a}, b = {b}. Добуток (тут сума): a + b = {a+b}")
print(f"3. Обернені елементи: a^-1 = {-a}, b^-1 = {-b}. Перевірка: a + (-a) = {a + (-a)} (нейтральний)")

order_Z10 = Z10.cardinality()
H_elements = sorted(list(set(Z10(i*a + j*b) for i in range(order_Z10) for j in range(order_Z10))))
print(f"4. Підгрупа <a, b>: {H_elements}. Порядок: {len(H_elements)}")

self_inv = [x for x in Z10 if -x == x]
print(f"5. Обернені самі до себе (x = -x): {self_inv}")
if Z10.is_commutative():
    print("6. Група є абелевою, тому a + b = b + a для всіх елементів.")
else:
    a_non, b_non = None, None
    for x in Z10:
        for y in Z10:
            if x + y != y + x:
                a_non, b_non = x, y
                break
        if a_non: break
    print(f"6. Група неабелева. Знайдено контрприклад: a = {a_non}, b = {b_non}. a+b = {a_non+b_non}, але b+a = {b_non+a_non}")
```

```sage
# ==========================================
# 2. Z_10^*
# ==========================================
print("=== 2. Група Z_10^* (Мультиплікативна) ===")
Z10_star = Integers(10).unit_group()
elems = Z10_star.list()
a, b = elems[1], elems[2]

print(f"2. Довільні елементи: a = {a.value()}, b = {b.value()}. Добуток: {(a*b).value()}")
print(f"3. Обернені елементи: a^-1 = {(a^-1).value()}, b^-1 = {(b^-1).value()}. Перевірка: a * a^-1 = {(a * a^-1).value()} (нейтральний)")

order_Z10_star = Z10_star.order()
H_vals = sorted(list(set(int(((a**i) * (b**j)).value()) for i in range(order_Z10_star) for j in range(order_Z10_star))))
print(f"4. Підгрупа <a, b>: {H_vals}. Порядок: {len(H_vals)}")

self_inv = [int(x.value()) for x in elems if x^-1 == x]
print(f"5. Обернені самі до себе (x^-1 = x): {self_inv}")

if Z10_star.is_abelian():
    print("6. Група є абелевою, тому a*b = b*a для всіх елементів.\n")
else:
    a_non, b_non = None, None
    for x in elems:
        for y in elems:
            if x * y != y * x:
                a_non, b_non = x, y
                break
        if a_non: break
    print(f"6. Група неабелева. Знайдено контрприклад: a = {a_non.value()}, b = {b_non.value()}. a*b = {(a_non*b_non).value()}, b*a = {(b_non*a_non).value()}\n")
```

```sage
# ==========================================
# 3. A_4
# ==========================================
print("=== 3. Група A_4 ===")
A4 = AlternatingGroup(4)
elems = A4.list()
a, b = elems[1], elems[2]

print(f"2. Довільні елементи: a = {a}, b = {b}. Добуток: {a*b}")
print(f"3. Обернені елементи: a^-1 = {a^-1}, b^-1 = {b^-1}. Перевірка: a * a^-1 = {a * a^-1} (нейтральний)")

H = A4.subgroup([a, b])
print(f"4. Підгрупа <a, b>: {H.list()}. Порядок: {H.order()}")

self_inv = [x for x in elems if x^-1 == x]
print(f"5. Обернені самі до себе (x^-1 = x): {self_inv}")

if A4.is_abelian():
    print("6. Група є абелевою, тому a*b = b*a для всіх елементів.")
else:
    a_non, b_non = None, None
    for x in elems:
        for y in elems:
            if x * y != y * x:
                a_non, b_non = x, y
                break
        if a_non: break
    print(f"6. Група неабелева. Знайдено контрприклад:\n   a = {a_non}\n   b = {b_non}\n   a*b = {a_non * b_non}\n   b*a = {b_non * a_non}")
```

```sage
# ==========================================
# 4. GL_2(Z_3)
# ==========================================
print("=== 4. Група GL_2(Z_3) ===")
GL2_3 = GL(2, GF(3))
elems_GL = GL2_3.list()

a = GL2_3([[2, 0], [0, 1]])
b = GL2_3([[1, 0], [0, 2]])

print(f"2. Довільні елементи:\na =\n{a}\nb =\n{b}\nДобуток a*b =\n{a*b}")

print(f"3. Обернені елементи:\na^-1 =\n{a^-1}\nb^-1 =\n{b^-1}\nПеревірка a * a^-1 =\n{a * a^-1} (одинична матриця)")

H = GL2_3.subgroup([a, b])
print(f"4. Порядок підгрупи <a, b>: {H.order()}")
print("   Елементи підгрупи:")
for matrix in H.list():
    print(f"{matrix}\n")

self_inv = [x for x in elems_GL if x^-1 == x]
print(f"5. Кількість елементів, обернених самі до себе: {len(self_inv)}")

if GL2_3.is_abelian():
    print("6. Група є абелевою, тому a*b = b*a для всіх матриць.\n")
else:
    a_non, b_non = None, None
    for x in elems_GL:
        for y in elems_GL:
            if x * y != y * x:
                a_non, b_non = x, y
                break
        if a_non: break
    print(f"6. Група неабелева. Знайдено контрприклад:\na =\n{a_non}\nb =\n{b_non}\na*b =\n{a_non*b_non}\nале b*a =\n{b_non*a_non}\n")
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
# ==========================================
# 1. Z_24
# ==========================================
print("=== Група Z_24 ===")
Z24 = Integers(24)
g_order = Z24.cardinality()

divisors = [d for d in range(1, 25) if 24 % d == 0]

subgroups_Z24 = []
for d in divisors:
    H_elements = [Z24(x) for x in range(0, 24, d)]
    gen = d if d != 24 else 0
    subgroups_Z24.append((gen, H_elements))

print(f"\n2. Список усіх підгруп (всього знайдено: {len(subgroups_Z24)}):")
for i, (gen, H) in enumerate(subgroups_Z24, 1):
    print(f"   {i}) Порядок {len(H):2} | Твірна: [{gen:2}] | Елементи: {H}")
    
print(f"3. Кількість підгруп з точністю до ізоморфізму: {len(subgroups_Z24)}")

orders_Z24 = sorted([len(H) for gen, H in subgroups_Z24])
print(f"4. Порядки всіх підгруп: {orders_Z24}")

check_Z24 = all(g_order % o == 0 for o in orders_Z24)
print(f"5. Чи ділить порядок кожної підгрупи порядок групи? {check_Z24}\n")
```

```sage
# ==========================================
# 2.Q_8
# ==========================================
print("=== Група Q_8 ===")
Q8 = QuaternionGroup()                                     
subgroups_Q8 = Q8.subgroups()
print(f"\n2. Список усіх підгруп (всього знайдено: {len(subgroups_Q8)}):")
for i, H in enumerate(subgroups_Q8, 1):
    print(f"   {i}) Порядок {H.order()} | Твірні: {H.gens()}")

iso_Q8 = []
for H in subgroups_Q8:
    if not any(H.is_isomorphic(K) for K in iso_Q8):
        iso_Q8.append(H)
print(f"\n3. Кількість підгруп з точністю до ізоморфізму: {len(iso_Q8)}")

orders_Q8 = sorted([H.order() for H in subgroups_Q8])
print(f"4. Порядки всіх підгруп: {orders_Q8}")

check_Q8 = all(Q8.order() % o == 0 for o in orders_Q8)
print(f"5. Чи ділить порядок кожної підгрупи порядок групи? {check_Q8}\n")
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
# ==========================================
# 1. D_4
# ==========================================
print("=== 1. Граф Келі для групи D_4 ===")
D4 = DihedralGroup(4)
gens_D4 = D4.gens()
C_D4 = D4.cayley_graph(generators=gens_D4)
print("1. Побудова графа Келі:")
print(f"   Твірні (S): {gens_D4}")
C_D4.plot().show()

v_D4 = C_D4.vertices()[0]
print(f"2. Степінь вершин: вихідна = {C_D4.out_degree(v_D4)}, вхідна = {C_D4.in_degree(v_D4)}")
print(f"3. Вершинно-транзитивний? {C_D4.is_vertex_transitive()}")
print(f"4. Гамільтонів цикл? {C_D4.is_hamiltonian()}\n")
```
