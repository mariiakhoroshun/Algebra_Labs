### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
for n in range(-10,11):
    if n>=0:
        if n==0:
            print(f"{n}=0")
        else:
            print(f"{n} додатнє")
    else:
        print(f"{n} від'ємне")
        
        
```

```sage
for n in range(1,11):
    print(n)
```

```sage
def check_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
 
for a in range(0,14):
    if check_prime(a):
        print(f"{a} просте")
```
