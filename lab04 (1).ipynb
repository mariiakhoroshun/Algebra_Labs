### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
G = SymmetricGroup(4)

H1 = G.subgroup([G('(1,2,3,4)')])
H2 = G.subgroup([G('(1,3)'), G('(1,2,3,4)')])
H3 = G.subgroup([G('(1,2)(3,4)'), G('(1,3)(2,4)')])

def analyze_subgroup(name, H):
    print(f"=== {name} ===")
    print(f"Елементи H: {list(H)}")
    
    order_H = H.order()
    index_H = G.order() // order_H
    print(f"1. |H| = {order_H}, [S_4 : H] = {index_H}")
    
    left_cosets = G.cosets(H, side='left')
    right_cosets = G.cosets(H, side='right')
    print("2. Ліві класи суміжності:")
    for i, c in enumerate(left_cosets, 1):
        print(f"   L{i}: {c}")
    print("   Праві класи суміжності:")
    for i, c in enumerate(right_cosets, 1):
        print(f"   R{i}: {c}")
        
    is_norm = H.is_normal(G)
    print(f"3. Нормальна підгрупа: {is_norm}")
    
    if is_norm:
        Q = G.quotient(H)
        ct = Q.cayley_table()
        print(f"4. Порядок факторгрупи S_4/H: {Q.order()}")
        print(f"   Абелева: {Q.is_abelian()}")
        print(f"   Структура: {Q.structure_description()}")
        print(f"   Ізоморфна S_3: {Q.is_isomorphic(SymmetricGroup(3))}")
        print(f"   Елементи таблиці Келі (a, b, c, ...): {list(ct.column_keys())}")
        print("   Таблиця Келі:")
        print(ct)
    else:
        a = next(g for g in G if set(g * h for h in H) != set(h * g for h in H))
        aH = [a * h for h in H]
        Ha = [h * a for h in H]
        print(f"5. Елемент a = {a}")
        print(f"   aH = {aH}")
        print(f"   Ha = {Ha}")
        print(f"   aH != Ha: {set(aH) != set(Ha)}")
```

```sage
analyze_subgroup("H_1 = <(1,2,3,4)>", H1)
```

```sage
analyze_subgroup("H_2 = <(1,3), (1,2,3,4)>", H2)
```

```sage
analyze_subgroup("H_3 = K_4", H3)
```

### 2. Нормальні підгрупи в діедральних групах

Розгляньте діедральні групи $D_n$ для значень $n\in\{3,4,\dots,30\}$ і виконайте такі кроки:
1. Побудуйте групу $D_n$.
2. Визначте кількість нормальних підгруп у $D_n$.
3. Складіть перелік порядків усіх нормальних підгруп.
4. Визначте алгебраїчну структуру кожної нормальної підгрупи.
5. Знайдіть закономірність і виведіть гіпотезу про кількість та структуру нормальних підгруп в $D_n$.
6. Сформулювати гіпотезу: Скільки нормальних підгруп в $D_n$ та яка їхня структура? Спрогнозувати кількість нормальних підгруп у $D_{470448}$ без виконання обчислень. 

```sage
results = {}

for n in range(3, 31):
    G = DihedralGroup(n)
    norms = sorted(G.normal_subgroups(), key=lambda H: (H.order(), H.structure_description()))
    orders = [H.order() for H in norms]
    structures = [H.structure_description() for H in norms]
    results[n] = (len(norms), orders, structures)
    print(f"n = {n:>2} (|D_{n}| = {G.order():>2}): кількість = {len(norms):>2}")
    print(f"   Порядки:   {orders}")
    print(f"   Структури: {structures}")
```

```sage
def predicted_normal_count(n):
    return number_of_divisors(n) + (3 if n % 2 == 0 else 1)

mismatches = [
    n for n, (actual_count, _, _) in results.items()
    if actual_count != predicted_normal_count(n)
]

print("Розбіжності з гіпотезою на n in [3..30]:", len(mismatches))

n_target = 470448
print(f"\nПрогноз для n = {n_target}:")
print(f"Розклад на прості множники: {factor(n_target)}")
print(f"Кількість дільників tau({n_target}) = {number_of_divisors(n_target)}")
print(f"Очікувана кількість нормальних підгруп у D_{n_target}: {predicted_normal_count(n_target)}")
```

```sage
def hyp_count(n):
    return number_of_divisors(n) + (3 if n % 2 == 0 else 1)

def check_hypothesis_for_n(n):
    G = DihedralGroup(n)
    actual = G.normal_subgroups()
    if len(actual) != hyp_count(n):
        return False
    expected = [CyclicPermutationGroup(d) for d in divisors(n)] + [DihedralGroup(n)]
    if n % 2 == 0:
        expected += [DihedralGroup(n // 2), DihedralGroup(n // 2)]
    unmatched = list(actual)
    for exp_H in expected:
        idx = next((i for i, H in enumerate(unmatched) if H.is_isomorphic(exp_H)), None)
        if idx is None:
            return False
        unmatched.pop(idx)
    return len(unmatched) == 0

test_range = range(3, 61)
failed = [n for n in test_range if not check_hypothesis_for_n(n)]
print(f"Перевірено n від 3 до 60 (кількість і структура всіх нормальних підгруп): розбіжностей = {len(failed)}")

n_big = 470448
tau_big = number_of_divisors(n_big)
total_big = hyp_count(n_big)
print(f"\nПрогноз для D_{n_big} без побудови групи:")
print(f"Розклад n = {n_big}: {factor(n_big)}")
print(f"Кількість дільників tau({n_big}) = {tau_big}")
print(f"Загальна кількість нормальних підгруп: {total_big}")
print(f"Структура: {tau_big} циклічних груп C_d (для d | {n_big}), 2 групи D_{n_big // 2} та 1 група D_{n_big}")
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
F = GF(3)
G = GL(2, F)

print("Порядок групи GL_2(Z_3):", G.order())
print("Твірні групи:")
for gen in G.gens():
    print(gen.matrix(), "\n")
```

```sage
V = VectorSpace(F, 2)
S4 = SymmetricGroup(4)

M = [
    [V([1, 0]), V([2, 0])],
    [V([0, 1]), V([0, 2])],
    [V([1, 1]), V([2, 2])],
    [V([1, 2]), V([2, 1])]
]

def line_to_set(line):
    return {tuple(v) for v in line}

M_sets = [line_to_set(l) for l in M]

def phi(g):
    A = g.matrix()
    perm = []
    for l in M:
        img = {tuple(v * A) for v in l}
        idx = M_sets.index(img) + 1
        perm.append(idx)
    return S4(perm)

print("Дія твірних на прямі з M:")
for gen in G.gens():
    print(gen.matrix())
    print("-> phi(g) =", phi(gen), "\n")
```

```sage
is_hom = all(phi(a * b) == phi(a) * phi(b) for a in G for b in G)
print("Перевірено всіх пар (a, b) з GL_2(Z_3) x GL_2(Z_3):", G.order()^2)
print("phi є гомоморфізмом груп:", is_hom)
```

```sage
ker_elems = [g for g in G if phi(g) == S4.identity()]
N = G.subgroup(ker_elems)

N_tuples = {tuple(h.matrix().list()) for h in N}
is_norm = all(tuple((g * h * g^(-1)).matrix().list()) in N_tuples for g in G for h in N)

print("Порядок ядра |N|:", N.order())
print("N є нормальною підгрупою в GL_2(Z_3):", is_norm)
print("Елементи ядра N:")
for g in N:
    print(g.matrix(), "\n")

Im = S4.subgroup([phi(g) for g in G])

print("Порядок образу |Im(phi)|:", Im.order())
print("Структура образу Im(phi):", Im.structure_description())
print("Im(phi) збігається з усією групою S_4:", Im == S4)
```

```sage
seen = set()
cosets = []
for g in G:
    c_set = frozenset(tuple((g * h).matrix().list()) for h in N)
    if c_set not in seen:
        seen.add(c_set)
        cosets.append((c_set, [g * h for h in N]))

coset_list = [c_set for c_set, _ in cosets]
S_Q = SymmetricGroup(len(coset_list))

def coset_perm(g):
    perm = [
        coset_list.index(frozenset(tuple((g * x).matrix().list()) for x in elems)) + 1
        for _, elems in cosets
    ]
    return S_Q(perm)

Q = S_Q.subgroup([coset_perm(g) for g in G])

print("Порядок факторгрупи GL_2(Z_3)/N:", Q.order())
print("Порядок образу Im(phi):", Im.order())
print("Структура факторгрупи:", Q.structure_description())
print("Структура образу:", Im.structure_description())
print("Факторгрупа ізоморфна образу (GL_2(Z_3)/N ≅ Im(phi)):", Q.is_isomorphic(Im))
```
