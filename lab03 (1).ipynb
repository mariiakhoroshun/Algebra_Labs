### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
n_values = [100, 173, 530, 746, 911, 1024]

for n in n_values:
    print("=" * 65)
    print(f"ГРУПА G = Z_{n} (порядок n = {n})")
    print("=" * 65)
    
    G = CyclicPermutationGroup(n)
    
    divs = divisors(n)
    print(f"1. Дільники n = {n}: {divs}")
    
    order_counts = {d: 0 for d in divs}
    for a in G:
        order_counts[a.order()] += 1
        
    print("2-3. Кількість елементів порядку d та phi(d):")
    for d in divs:
        phi_d = euler_phi(d)
        print(f"   d = {d:<4} | к-сть = {order_counts[d]:<4} | phi(d) = {phi_d:<4} | збіг: {order_counts[d] == phi_d}")
        
    subgroups = sorted(G.subgroups(), key=lambda H: H.order())
    sub_list = [f"<{ (n // H.order()) % n }> (пор. {H.order()})" for H in subgroups]
    print(f"4. Усі підгрупи Z_{n}:\n   " + ", ".join(sub_list))

    num_sub = len(subgroups)
    tau_n = number_of_divisors(n)
    print(f"5-6. Кількість підгруп: {num_sub} | tau({n}) = {tau_n} | збіг: {num_sub == tau_n}\n")
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
n_list = [5, 6, 10, 12, 16, 17, 20]

for n in n_list:
    D = DihedralGroup(n)
    order_D = D.order()
    divs = divisors(order_D)
    
    counts = {d: 0 for d in divs}
    for g in D.list():
        counts[g.order()] += 1
        
    print("=" * 65)
    print(f"Діедральна група D_{n} (порядок |D_{n}| = {order_D})")
    print(f"1. Кількість елементів порядку d = 2: {counts[2]}")
    print("2. Кількість елементів для кожного дільника d > 2 числа 2n:")
    for d in divs:
        if d > 2:
            theoretical = euler_phi(d) if (n % d == 0) else 0
            print(f"   d = {d:<3} | кількість = {counts[d]:<3} "
                  f"(теоретично: {theoretical}, d ділить n? {n % d == 0})")
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
n = 1102158078129785185997827239806

fact_n = factor(n)
print(f"1. Факторизація n:\n   n = {fact_n}")

phi_n = euler_phi(n)

try:
    gen = primitive_root(n)
    is_cyclic = True
except ValueError:
    is_cyclic = False

print(f"\n2. Чи є група Z_n^* циклічною? {is_cyclic}")
print(f"   Порядок групи |Z_n^*| = phi(n) = {phi_n}")
print(f"   Факторизація порядку phi(n) = {factor(phi_n)}")

R = Zmod(n)
phi_prime_divs = prime_divisors(phi_n)

a_min = None
for a_cand in range(2, 1000):
    if gcd(a_cand, n) == 1:
        el = R(a_cand)
        if all(el**(phi_n // q) != 1 for q in phi_prime_divs):
            a_min = a_cand
            break

a_bar = R(a_min)
print(f"\n3. Найменше натуральне число a (твірний елемент): a = {a_min}")

ord_a = a_bar.multiplicative_order()
print(f"\n4. Порядок елемента a_bar: ord({a_min}) = {ord_a}")
print(f"   Перевірка ord(a_bar) == phi(n): {ord_a == phi_n}")

set_random_seed(42)
random_elements = []
while len(random_elements) < 3:
    elem = R.random_element()
    if gcd(lift(elem), n) == 1 and elem not in random_elements:
        random_elements.append(elem)

g1, g2, g3 = random_elements
print("\n5. Три випадкові елементи з Z_n^*:")
print(f"   g_1 = {g1}")
print(f"   g_2 = {g2}")
print(f"   g_3 = {g3}")

print("\n6. Дискретні логарифми за основою a_bar:")
for i, g in enumerate([g1, g2, g3], 1):
    k = discrete_log(g, a_bar, ord=phi_n)
    check = (a_bar**k == g)
    print(f"   Для g_{i} = {g}:")
    print(f"   k_{i} = log_{a_min}(g_{i}) = {k}")
    print(f"   Перевірка (a_bar^k_{i} == g_{i}): {check}\n")
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:
1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?

```sage
A = matrix(ZZ, [[1, 0, 1],
                [0, 1, 0],
                [0, 0, 1]])

B = matrix(ZZ, [[1, 0, 0],
                [0, 1, 1],
                [0, 0, 1]])

G = MatrixGroup([A, B])

print("Твірна матриця A:\n", A)
print("Твірна матриця B:\n", B)

print("\n1. Чи є група G скінченною?")
print("   G.is_finite() =", G.is_finite())
print("   Порядок групи G.order() =", G.order())

print("\n2. Чи є група G абелевою?")
print("   Чи комутують твірні (A * B == B * A)?", A * B == B * A)
print("   G.is_abelian() =", G.is_abelian())

var('k m n', domain='integer')
A_sym = matrix(SR, [[1, 0, k], [0, 1, 0], [0, 0, 1]])
B_sym = matrix(SR, [[1, 0, 0], [0, 1, m], [0, 0, 1]])
g_km = A_sym * B_sym

print("\n4. Загальний вигляд елемента g(k, m) = A^k * B^m:")
print(g_km)

var('k1 m1 k2 m2', domain='integer')
g1 = matrix(SR, [[1, 0, k1], [0, 1, m1], [0, 0, 1]])
g2 = matrix(SR, [[1, 0, k2], [0, 1, m2], [0, 0, 1]])
print("\n   Добуток двох елементів g(k1, m1) * g(k2, m2):")
print(g1 * g2)
print("\n   Обернений елемент g(k, m)^(-1):")
print(g_km.inverse())

print("\n3. Перевірка порядків конкретних елементів у SageMath:")
print("   Порядок одиничної матриці G.one().order() =", G.one().order())
print("   Порядок матриці A:", A.multiplicative_order())
print("   Порядок матриці B:", B.multiplicative_order())

elem = (A^3 * B^(-5)).change_ring(ZZ)
print("   Порядок A^3 * B^(-5):", elem.multiplicative_order())

Z2 = ZZ^2
vA = vector(ZZ, [A[0, 2], A[1, 2]])
vB = vector(ZZ, [B[0, 2], B[1, 2]])
H_iso = Z2.submodule([vA, vB])

print("\n5-6. Циклічність та ізоморфізм:")
print("   Образи твірних A і B у Z^2:", vA, "та", vB)
print("   Ранг групи G (кількість незалежних твірних):", H_iso.rank())
print("   Чи є група G циклічною (ранг == 1)?", H_iso.rank() == 1)
print("   Чи ізоморфна група G всій групі Z x Z (H_iso == Z^2)?", H_iso == Z2)
```
